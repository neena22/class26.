# Lesson Summary — Git & Collaboration

This notebook summarizes the Pull Request workflow:

1. Create a new branch from `main`.
2. Open the notebook from the new branch in Google Colab.
3. Add or edit one cell.
4. Save the change as a commit on the same branch.
5. Open a Pull Request into `main`.
6. Review the diff and merge the Pull Request.

**Change made:** Added this Markdown summary cell.

**Out of scope:** Changing existing code cells or adding new features.

**Reviewer check:** Confirm that only the intended summary cell changed.
